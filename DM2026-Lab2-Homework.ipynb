# Project Report

## 1. Model Development

The objective of this project is to classify short social-media posts into six emotion categories: **anger, disgust, fear, joy, sadness, and surprise**.

I experimented with traditional text representations, sentence embeddings, pretrained emotion models, and probability-based model fusion. The selected competition model was **Model O: Neutral-Aware Fusion**, which combines a TF-IDF + MiniLM Logistic Regression classifier with a pretrained Emotion RoBERTa model.

### 1.1 Preprocessing Steps

**1. Data loading and integration**

The competition data was loaded from four files: `data_identification.csv`, `emotion.csv`, `final_posts.json`, and `sampleSubmission.csv`.

The datasets were merged using `post_id`, resulting in:

- Training set: **3,360 labeled samples**
- Test set: **1,440 unlabeled samples**
- Target classes: anger, disgust, fear, joy, sadness, and surprise

**2. Data quality checks**

I checked the dataset structure, missing values, class distribution, and duplicate text records.

The training data contained 50 duplicate text occurrences beyond the first occurrence of each text, including 16 groups with conflicting emotion labels.

I also identified 59 test samples whose normalized text matched a training sample. All 59 matched texts had a single unique training label, although the model disagreed with that label on seven samples.

These findings showed that duplicate text and label ambiguity needed to be considered during validation.

**3. Text preprocessing**

I retained the original text for the pretrained language models because capitalization, punctuation, and short expressions can provide important emotional context.

The TF-IDF branch used word-level and character-level representations to extract lexical patterns, while MiniLM and RoBERTa processed the original text through their respective tokenizers.

**4. Validation strategy**

I used `StratifiedGroupKFold` with five splits, shuffling enabled, and `random_state=42`.

Identical text strings were assigned to the same group so they would not appear in both the training and validation partitions of a fold.

I evaluated models using **Accuracy, Macro F1, and per-class precision, recall, and F1**. Macro F1 was especially useful because the six emotion classes had different sample counts.

### 1.2 Feature Engineering Steps

I investigated three complementary representations of emotion-related text.

**1. Word and Character TF-IDF**

Word-level TF-IDF captures important vocabulary and phrases, while character-level TF-IDF captures subword patterns, spelling variations, and informal expressions.

This representation is computationally efficient and works well with linear classifiers.

**2. MiniLM Sentence Embeddings**

I used the pretrained `sentence-transformers/all-MiniLM-L6-v2` model to generate **384-dimensional normalized sentence embeddings**.

Unlike TF-IDF, MiniLM represents the meaning of a sentence rather than relying only on exact word matches.

**3. Emotion RoBERTa Probabilities**

I used the pretrained `j-hartmann/emotion-english-distilroberta-base` model to estimate probabilities for seven emotions:

anger, disgust, fear, joy, neutral, sadness, and surprise.

The competition requires only six classes. Instead of immediately removing and renormalizing the neutral category, my final approach preserved its probability mass when combining RoBERTa with the Logistic Regression model.

**4. Feature Fusion**

For the Logistic Regression branch, I concatenated the sparse TF-IDF feature matrix with the MiniLM embeddings.

The final classifier was trained using these combined features, while RoBERTa provided an additional set of emotion probabilities for the final fusion stage.

### 1.3 Explanation of Your Model

**1. Classification architecture**

The selected model consists of two branches.

**Branch A — TF-IDF + MiniLM + Logistic Regression**

- TF-IDF extracts word-level and character-level patterns.
- MiniLM produces semantic sentence embeddings.
- Logistic Regression predicts the six competition emotions.

The Logistic Regression classifier used `C=1.0`, the `lbfgs` solver, and a maximum of 2,000 iterations.

**Branch B — Pretrained Emotion RoBERTa**

- RoBERTa analyzes contextual emotional information.
- Its output contains six target emotion probabilities plus neutral.
- The pretrained RoBERTa parameters were kept frozen.

**2. Neutral-Aware Fusion**

The two branches were combined using equal weights:

\[
s_c = 0.5P_K(c) + 0.5P_R(c)
\]

Here, \(P_K(c)\) is the Logistic Regression probability for emotion \(c\), and \(P_R(c)\) is the original, unnormalized RoBERTa probability for that emotion, before removing neutral.

The six combined scores were subsequently normalized, and the class with the highest score was selected.

This method preserves the influence of RoBERTa's neutral probability instead of forcing its six non-neutral probabilities to sum to one before fusion.

**3. Model comparison**

The following results were obtained from local exploratory evaluation:

| Model | Accuracy | Macro F1 |
|---|---:|---:|
| Model K: TF-IDF + MiniLM + Logistic Regression | 0.5744 | 0.5518 |
| Model M: Pretrained Emotion RoBERTa | 0.5768 | 0.5715 |
| Model N: Equal-weight probability fusion | 0.6113 | 0.6006 |
| **Model O: Neutral-Aware Fusion** | **0.6131** | **0.6026** |
| Model P: Learned feature fusion | 0.6098 | 0.5982 |

Model O achieved the highest observed Accuracy and Macro F1 among these evaluated configurations.

Its F1 score for the difficult **disgust** class was approximately **0.4869**. Disgust remained one of the most challenging classes due to its semantic overlap with anger and sadness.

**4. Final model selection**

I selected Model O because it provided the strongest observed overall balance across the six emotion categories.

Model P was not selected because, although it improved performance on anger and fear, it reduced overall Macro F1 and performed worse on disgust, joy, and sadness.

The local evaluation results should be interpreted as exploratory estimates because several model configurations were compared on the same evaluation framework. They are not equivalent to the official Kaggle Private Leaderboard score.

**5. Kaggle prediction output**

For final inference, the Logistic Regression branch was trained on all 3,360 labeled samples and combined with pretrained RoBERTa predictions.

The resulting CSV contained **1,440 predictions** with exactly two columns: `id` and `label`.

I verified the uniqueness and ordering of submission IDs and checked that all predictions belonged to the six allowed classes.

````python
import json
import pandas as pd
from pathlib import Path

# 1. Dataset folder
data_dir = Path("kaggle_data")

print("=== Loading Competition Data ===")
print("Current folder:", Path.cwd())

# 2. Read CSV files
identification = pd.read_csv(
    data_dir / "data_identification.csv"
)

emotion = pd.read_csv(
    data_dir / "emotion.csv"
)

sample = pd.read_csv(
    data_dir / "sampleSubmission.csv"
)

# 3. Read text data
with open(
    data_dir / "final_posts.json",
    "r",
    encoding="utf-8"
) as f:
    raw_posts = json.load(f)

# Support common JSON formats
if isinstance(raw_posts, list):
    posts = pd.DataFrame(raw_posts)

elif isinstance(raw_posts, dict):
    if "post_id" in raw_posts and "text" in raw_posts:
        posts = pd.DataFrame(raw_posts)
    else:
        posts = pd.DataFrame([
            {
                "post_id": str(post_id),
                "text": (
                    value.get("text")
                    if isinstance(value, dict)
                    else value
                )
            }
            for post_id, value in raw_posts.items()
        ])
else:
    raise ValueError("Unexpected final_posts.json format")

print("Posts columns:", posts.columns.tolist())

assert {"post_id", "text"}.issubset(posts.columns)

# 4. Merge datasets by post_id
dataset = identification.merge(
    posts[["post_id", "text"]],
    on="post_id",
    how="left",
    validate="one_to_one"
)

dataset = dataset.merge(
    emotion,
    on="post_id",
    how="left",
    validate="one_to_one"
)

# 5. Build training dataframe
train_df = dataset.loc[
    dataset["set"] == "train",
    ["post_id", "text", "label"]
].copy().reset_index(drop=True)

# 6. Build test dataframe in submission order
test_df = sample[["post_id"]].merge(
    dataset.loc[
        dataset["set"] == "test",
        ["post_id", "text"]
    ],
    on="post_id",
    how="left",
    validate="one_to_one"
)

# 7. Data validation
assert len(train_df) == 3360
assert len(test_df) == 1440

assert train_df["text"].notna().all()
assert train_df["label"].notna().all()
assert test_df["text"].notna().all()

print("\n=== Dataset Summary ===")
print("Training shape:", train_df.shape)
print("Test shape:", test_df.shape)

print("\n=== Emotion Distribution ===")
print(train_df["label"].value_counts())

print("\n=== Training Preview ===")
display(train_df.head())

print("\n=== Test Preview ===")
display(test_df.head())

print("\nDATA PREPROCESSING COMPLETED")
````

````python
import pandas as pd
from sklearn.model_selection import StratifiedGroupKFold

# Use identical text as the same group
train_df["text_group"] = (
    train_df["text"].astype(str).str.strip()
)

# Check whether identical texts have different labels
label_counts = (
    train_df.groupby("text_group")["label"]
    .nunique()
)

conflicting_groups = (label_counts > 1).sum()

print("=== Duplicate Label Check ===")
print("Duplicate text rows:",
      train_df["text_group"].duplicated().sum())

print("Text groups with conflicting labels:",
      conflicting_groups)

# Create a group-aware validation split
splitter = StratifiedGroupKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

train_indices, val_indices = next(
    splitter.split(
        X=train_df,
        y=train_df["label"],
        groups=train_df["text_group"]
    )
)

model_train_df = train_df.iloc[train_indices].copy()
val_df = train_df.iloc[val_indices].copy()

# Check for identical texts in both groups
train_texts = set(model_train_df["text_group"])
val_texts = set(val_df["text_group"])

overlap = train_texts & val_texts

print("\n=== Validation Split ===")
print("Training samples:", len(model_train_df))
print("Validation samples:", len(val_df))
print("Overlapping text groups:", len(overlap))

assert len(overlap) == 0, "Text leakage detected!"

# Check class distributions
print("\n=== Class Distribution ===")

distribution = pd.DataFrame({
    "Training": model_train_df["label"].value_counts(),
    "Validation": val_df["label"].value_counts()
}).fillna(0).astype(int)

display(distribution)

# Prepare inputs for the classifier
X_train = model_train_df["text"]
y_train = model_train_df["label"]

X_val = val_df["text"]
y_val = val_df["label"]

print("\nVALIDATION SPLIT COMPLETED")
````

## 2. Feature Engineering Steps

````python
import re
import numpy as np
import pandas as pd

from pathlib import Path
from scipy.sparse import csr_matrix, hstack
from sklearn.base import clone
from sklearn.pipeline import Pipeline, FeatureUnion
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import LinearSVC

from sentence_transformers import SentenceTransformer
from transformers import pipeline

print("=== MODEL O FEATURE ENGINEERING ===")

# ============================================================
# 1. Verify preprocessing results
# ============================================================

assert "train_df" in globals(), (
    "Run Preprocessing Steps first."
)

assert "test_df" in globals(), (
    "Run Preprocessing Steps first."
)

assert {"text", "label"}.issubset(train_df.columns)
assert {"post_id", "text"}.issubset(test_df.columns)

assert len(train_df) == 3360
assert len(test_df) == 1440

labels = sorted(train_df["label"].unique())

assert labels == sorted([
    "anger",
    "disgust",
    "fear",
    "joy",
    "sadness",
    "surprise"
])

print("Preprocessing: READY")

# ============================================================
# 2. Restore original Model E feature definitions
# ============================================================

def make_char_features():
    return TfidfVectorizer(
        analyzer="char_wb",
        ngram_range=(3, 6),
        min_df=2,
        max_features=30000,
        sublinear_tf=True
    )


LOW_INFORMATION_UNIGRAMS = {
    "a", "an", "the", "and", "will"
}


def emotion_word_analyzer(text):

    text = str(text).lower().replace("’", "'")

    tokens = re.findall(
        r"[a-z]+(?:'[a-z]+)?|\d+",
        text
    )

    features = []

    # Keep informative unigrams
    for token in tokens:
        if token not in LOW_INFORMATION_UNIGRAMS:
            features.append(token)

    # Keep contextual bigrams
    for i in range(len(tokens) - 1):
        features.append(
            tokens[i] + " " + tokens[i + 1]
        )

    return features


def make_svm():
    return LinearSVC(
        C=0.3,
        random_state=42,
        max_iter=5000
    )


# Original Model E from Step 35B
model_e = Pipeline([
    ("features", FeatureUnion([
        (
            "char",
            make_char_features()
        ),
        (
            "word",
            TfidfVectorizer(
                analyzer=emotion_word_analyzer,
                min_df=2,
                max_features=20000,
                sublinear_tf=True
            )
        )
    ])),
    (
        "classifier",
        make_svm()
    )
])

print("Original Model E: RESTORED")

# ============================================================
# 3. Load Kaggle submission order
# ============================================================

template_path = Path(
    "kaggle_data/sampleSubmission.csv"
)

assert template_path.exists(), (
    f"Missing submission template: {template_path}"
)

template = pd.read_csv(
    template_path,
    dtype=str,
    keep_default_na=False
)

template.columns = template.columns.str.strip()

template_id_col = next(
    (
        c for c in ["post_id", "id"]
        if c in template.columns
    ),
    None
)

assert template_id_col is not None

template_ids = (
    template[template_id_col]
    .astype(str)
    .str.strip()
)

lookup = test_df[["post_id", "text"]].copy()

lookup["match_id"] = (
    lookup["post_id"]
    .astype(str)
    .str.strip()
)

assert template_ids.is_unique
assert lookup["match_id"].is_unique

ordered_test = pd.DataFrame({
    "match_id": template_ids
}).merge(
    lookup[["match_id", "text"]],
    on="match_id",
    how="left",
    validate="one_to_one",
    sort=False
)

assert len(ordered_test) == 1440
assert ordered_test["text"].notna().all()

print("Submission IDs: VERIFIED")

# ============================================================
# 4. Prepare full training and test text
# ============================================================

X_train = (
    train_df["text"]
    .astype(str)
    .reset_index(drop=True)
)

y_train = (
    train_df["label"]
    .astype(str)
    .reset_index(drop=True)
)

X_test = (
    ordered_test["text"]
    .astype(str)
    .reset_index(drop=True)
)

print("Training samples:", len(X_train))
print("Test samples:", len(X_test))

# ============================================================
# 5. Word + Character TF-IDF
# ============================================================

vectorizer_o = clone(
    model_e.named_steps["features"]
)

train_tfidf = vectorizer_o.fit_transform(
    X_train,
    y_train
)

test_tfidf = vectorizer_o.transform(
    X_test
)

assert train_tfidf.shape[1] == test_tfidf.shape[1]

print("\n=== TF-IDF ===")
print("Train:", train_tfidf.shape)
print("Test:", test_tfidf.shape)

# ============================================================
# 6. MiniLM Encoder
# ============================================================

encoder = SentenceTransformer(
    "sentence-transformers/all-MiniLM-L6-v2",
    device="cpu"
)

print("\n=== MiniLM Embeddings ===")

minilm_embeddings = encoder.encode(
    X_train.tolist(),
    batch_size=32,
    show_progress_bar=True,
    convert_to_numpy=True,
    normalize_embeddings=True
)

train_minilm = np.asarray(
    minilm_embeddings,
    dtype=np.float32
)

test_minilm = encoder.encode(
    X_test.tolist(),
    batch_size=32,
    show_progress_bar=True,
    convert_to_numpy=True,
    normalize_embeddings=True
)

test_minilm = np.asarray(
    test_minilm,
    dtype=np.float32
)

assert train_minilm.shape == (3360, 384)
assert test_minilm.shape == (1440, 384)

assert np.isfinite(train_minilm).all()
assert np.isfinite(test_minilm).all()

print("Train:", train_minilm.shape)
print("Test:", test_minilm.shape)

# ============================================================
# 7. Combine TF-IDF + MiniLM
# ============================================================

train_features_o = hstack([
    train_tfidf,
    csr_matrix(train_minilm)
], format="csr")

test_features_o = hstack([
    test_tfidf,
    csr_matrix(test_minilm)
], format="csr")

assert train_features_o.shape[1] == (
    test_features_o.shape[1]
)

print("\n=== Combined Features ===")
print("Train:", train_features_o.shape)
print("Test:", test_features_o.shape)

# ============================================================
# 8. Load pretrained Emotion RoBERTa
# ============================================================

emotion_classifier = pipeline(
    task="text-classification",
    model="j-hartmann/emotion-english-distilroberta-base",
    device=-1
)

print("\n=== Emotion RoBERTa ===")
print("Model: READY")

# ============================================================
# 9. Extract seven-class Emotion RoBERTa scores
# ============================================================

m_raw_o = np.zeros(
    (len(X_test), 6),
    dtype=np.float64
)

neutral_o = np.zeros(
    len(X_test),
    dtype=np.float64
)

batch_size = 16

for start in range(0, len(X_test), batch_size):

    end = min(
        start + batch_size,
        len(X_test)
    )

    results = emotion_classifier(
        X_test.iloc[start:end].tolist(),
        top_k=None,
        truncation=True,
        max_length=256,
        batch_size=batch_size
    )

    for j, result in enumerate(results):

        score_map = {
            item["label"].lower(): float(item["score"])
            for item in result
        }

        idx = start + j

        m_raw_o[idx] = [
            score_map[label]
            for label in labels
        ]

        neutral_o[idx] = score_map["neutral"]

    if end % 160 == 0 or end == len(X_test):
        print(
            f"RoBERTa processed: {end}/{len(X_test)}"
        )

# ============================================================
# 10. Validate probabilities and feature dimensions
# ============================================================

assert m_raw_o.shape == (1440, 6)
assert neutral_o.shape == (1440,)

assert np.isfinite(m_raw_o).all()
assert np.isfinite(neutral_o).all()

assert np.allclose(
    m_raw_o.sum(axis=1) + neutral_o,
    1.0,
    atol=1e-4
)

expected_shape_train = (3360, 38231)
expected_shape_test = (1440, 38231)

assert train_features_o.shape == expected_shape_train, (
    "Feature dimensions differ from original Model O."
)

assert test_features_o.shape == expected_shape_test, (
    "Test feature dimensions differ from original Model O."
)

# ============================================================
# 11. Final verification
# ============================================================

print("\n===================================")
print("FEATURE ENGINEERING COMPLETED")
print("===================================")

print("Model E: READY")
print("MiniLM: READY")
print("Emotion RoBERTa: READY")

print("\nTrain features:", train_features_o.shape)
print("Test features:", test_features_o.shape)

print("RoBERTa scores:", m_raw_o.shape)
print("Neutral scores:", neutral_o.shape)

print("Matched IDs:", len(ordered_test))

print("\nALL MODEL O FEATURES READY")
````

## 3. Model Implementation Steps

````python
from sklearn.linear_model import LogisticRegression

# 1. Train Model K on all training data
final_model_k = LogisticRegression(
    C=1.0,
    solver="lbfgs",
    max_iter=2000,
    random_state=42
)

final_model_k.fit(
    train_features_o,
    y_train
)

# 2. Get six-class probability predictions
class_indices = [
    list(final_model_k.classes_).index(label)
    for label in labels
]

k_prob_o = final_model_k.predict_proba(
    test_features_o
)[:, class_indices]

# 3. Model O Neutral-Aware Fusion
# Do NOT renormalize RoBERTa before blending
fusion_scores_o = (
    0.5 * k_prob_o
    + 0.5 * m_raw_o
)

fusion_probs_o = (
    fusion_scores_o
    / fusion_scores_o.sum(axis=1, keepdims=True)
)

predictions_o = np.asarray(labels)[
    fusion_probs_o.argmax(axis=1)
]

# 4. Generate official Kaggle CSV
submission_o = pd.DataFrame({
    "id": ordered_test["match_id"].to_numpy(),
    "label": predictions_o
})

valid_labels = set(labels)

assert submission_o.columns.tolist() == [
    "id", "label"
]

assert len(submission_o) == 1440
assert submission_o["id"].is_unique
assert submission_o["label"].isin(valid_labels).all()

assert (
    submission_o["id"].tolist()
    == template_ids.tolist()
)

submission_o.to_csv(
    "submission_model_o_best.csv",
    index=False,
    encoding="utf-8"
)

# 5. Verify saved CSV
verified = pd.read_csv(
    "submission_model_o_best.csv",
    dtype=str
)

assert verified.equals(submission_o)

print("=== MODEL O READY ===")
print("Rows:", len(verified))
print("Columns:", verified.columns.tolist())

print("\nPrediction Distribution:")
display(verified["label"].value_counts())

print("\nSubmission Preview:")
display(verified.head(10))

print("\nFile: submission_model_o_best.csv")
print("KAGGLE SUBMISSION VERIFIED")
````
